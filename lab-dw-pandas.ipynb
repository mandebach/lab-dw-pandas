### Data Description

- Customer - Customer ID

- ST - State where customers live

- Gender - Gender of the customer

- Education - Background education of customers 

- Customer Lifetime Value - Customer lifetime value(CLV) is the total revenue the client will derive from their entire relationship with a customer. In other words, is the predicted or calculated value of a customer over their entire duration as a policyholder with the insurance company. It is an estimation of the net profit that the insurance company expects to generate from a customer throughout their relationship with the company. Customer Lifetime Value takes into account factors such as the duration of the customer's policy, premium payments, claim history, renewal likelihood, and potential additional services or products the customer may purchase. It helps insurers assess the long-term profitability and value associated with retaining a particular customer.

- Income - Customers income

- Monthly Premium Auto - Amount of money the customer pays on a monthly basis as a premium for their auto insurance coverage. It represents the recurring cost that the insured person must pay to maintain their insurance policy and receive coverage for potential damages, accidents, or other covered events related to their vehicle.

- Number of Open Complaints - Number of complaints the customer opened

- Policy Type - There are three type of policies in car insurance (Corporate Auto, Personal Auto, and Special Auto)

- Vehicle Class - Type of vehicle classes that customers have Two-Door Car, Four-Door Car SUV, Luxury SUV, Sports Car, and Luxury Car

- Total Claim Amount - the sum of all claims made by the customer. It represents the total monetary value of all approved claims for incidents such as accidents, theft, vandalism, or other covered events.

```python
import pandas as pd
import numpy as np

customer_df = pd.read_csv("file1.csv")

# Identify how many rows and columns and data types => 4008 rows x 11 columns
customer_df.info()
```

```python
# Check how many empty rows there are - mismatch between 1071 non-null entries and 4008 rows
empty_rows = customer_df.isna().all(axis=1).sum()

print(empty_rows)

# Drop the empty rows
customer_df = customer_df.dropna(how="all")

customer_df.info()

```

```python
# Clean up State column
customer_df["ST"].value_counts(dropna=False)
customer_df["ST"] = customer_df["ST"].replace({
    "Cali": "California",
    "WA": "Washington",
    "AZ": "Arizona"
})
customer_df["ST"].value_counts(dropna=False)
```

```python
# Change Gender to be a category. Clean up the data.
customer_df["GENDER"].value_counts(dropna=False)

customer_df["GENDER"] = customer_df["GENDER"].replace({
    "Male": "M",
    "Femal": "F",
    "female": "F"
})

customer_df["GENDER"].value_counts(dropna=False)
customer_df["GENDER"].value_counts(normalize=True) * 100
```

```python
# Clean Customer Lifetime Value
# Check a sample of the data
customer_df["Customer Lifetime Value"].sample(10)

# See if they all end with %
customer_df["Customer Lifetime Value"].str.endswith("%").value_counts()
```

```python
# Remove % for each entry - rstrip removes it from the right. This also converts it to float since this is a currency.
customer_df["Customer Lifetime Value"] = (
    customer_df["Customer Lifetime Value"]
    .str.rstrip("%")
    .astype(float)
)
```

```python
# These numbers feel too large for CLV. Since they used to have a percentage. I think I need to divide by 100

customer_df["Customer Lifetime Value"].describe()
customer_df["Customer Lifetime Value"] = customer_df["Customer Lifetime Value"] / 100
```

```python
customer_df["Customer Lifetime Value"].describe()
```

```python
customer_df.nunique()
```

```python
# Clean up the Education column
customer_df["Education"].value_counts(dropna=False)
customer_df["Education"] = customer_df["Education"].replace({"Bachelors": "Bachelor"})
customer_df["Education"].value_counts(normalize=True) * 100
```

```python
customer_df["Income"].sample(10)
customer_df["Income"].describe()
```

```python
customer_df["Monthly Premium Auto"].sample(10)
customer_df["Monthly Premium Auto"].describe()
```

```python
# Clean up the Complaints column

customer_df["Number of Open Complaints"].sample(10)
```

```python
# This looks like it has incorrectly been stored as a date.

customer_df["Number of Open Complaints"].value_counts()
```

```python
customer_df["Number of Open Complaints"] = (
    customer_df["Number of Open Complaints"]
    .str.split("/")   # strips away the / and makes a list of 3 separate numbers
    .str[1]           # takes the second indexed number
    .astype(int)      # converts to an integer
)
customer_df["Number of Open Complaints"].describe()
```

```python
# Policy Type column looks fine
customer_df["Policy Type"].value_counts()
```

```python
# Vehicle Class column looks fine
customer_df["Vehicle Class"].value_counts()
```

```python
# Total Claim Amount
customer_df["Total Claim Amount"].value_counts()

# Convert to Float
customer_df["Total Claim Amount"] = (customer_df["Total Claim Amount"].astype(float))  # converts to Float

customer_df["Total Claim Amount"].describe()
```
